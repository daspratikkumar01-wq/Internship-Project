Q31. Find all Armstrong numbers between 1 and 9999. (An n-digit number is Armstrong if the sum of its digits each raised to the power n equals the number itself. Example: 153 = 1³ + 5³ + 3³.) Print them in a clean comma-separated list with a count at the end.

```python
armstrong_numbers = []
for num in range(1, 10000):
  temp = num
  digits = 0
  while temp > 0:
    digits += 1
    temp = temp // 10
  temp = num
  total = 0
  while temp > 0:
    digit = temp % 10
    total = total + digit ** digits
    temp = temp // 10
  if total == num:
    armstrong_numbers.append(num)
print("Armstrong numbers:", end=" ")
for i in range(len(armstrong_numbers)):
  if i < len(armstrong_numbers) - 1:
    print(armstrong_numbers[i], end=", ")
  else:
    print(armstrong_numbers[i])
print("Count:", len(armstrong_numbers))
```

Q33. Prime check, optimized. Take a number N. Check if it is prime by iterating only up to √N. Then, print the SUM of all primes between 2 and N (inclusive). Do not use any library.

```python
n = int(input("Enter a number:"))
if n < 2:
  print(n, "is not a Prime number")
else:
  is_prime = True
  for i in range(2, n):
    if i * i > n:
      break
    if n % i == 0:
      is_prime = False
      break
  if is_prime:
    print(n, "is a prime number")
  else:
    print(n, "is not a prime number")
prime_sum = 0
for num in range(2, n+1):
  prime = True
  for j in range(2, num):
    if j * j > num:
      break
    if num % j == 0:
      prime = False
      break
  if prime:
    prime_sum = prime_sum + num
print("Sum of prime numbers:", prime_sum)
```

Q38. Word Frequency Analyzer for Swiggy reviews. Take a review sentence. Find the frequency of each word (case-insensitive, "Food" and "food" should be counted together). Print words sorted by frequency in descending order. Also print the MOST frequent word and its count. Use a dict.

```python
review = input("Enter a review:")
words = review.lower().split()
frequency = {}
for word in words:
  if word in frequency:
    frequency[word] += 1
  else:
    frequency[word] = 1
sorted_words = sorted(frequency, key = frequency.get, reverse = True)
print("Word Frequencies:")
for word in sorted_words:
  print(word, ":", frequency[word])
most_frequent = sorted_words[0]
print("Most frequent word:", most_frequent)
print("Count:", frequency[most_frequent])
```

```python
from enum import unique
section_a = {"Python", "Java", "C++", "SQL"}
section_b = {"Java", "JavaScript", "Python", "Go", "SQL"}
both = section_a & section_b
only_a = section_a - section_b
only_b = section_b - section_a
unique = section_a | section_b
exclusive = section_a ^ section_b
print("Common languages:", both)
print("Only Section A:", only_a)
print("Only Section B:", only_b)
print("Total unique languages:", unique)
print("Symmetric difference:", exclusive)
```

Q41. TCS stock prices for a week stored as a tuple: `prices = (3845, 3902, 3878, 3955, 4012, 3998, 4055)` (Mon to Sun in ₹). Find: (a) maximum price and which day, (b) minimum price and which day, (c) average, (d) biggest single-day gain (and between which days), (e) total volatility = sum of absolute day-to-day price changes.

```python
prices = (3845, 3902, 3878, 3955, 4012, 3998, 4055)
days = ("Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday")
maximum = max(prices)
minimum = min(prices)
max_day = days[prices.index(maximum)]
min_day = days[prices.index(minimum)]
average = sum(prices) / len(prices)
biggest_gain = 0
gain_days = ""
for i in range(1, len(prices)):
  gain = prices[i] - prices[i-1]
  if gain > biggest_gain:
    biggest_gain = gain
    gain_days = days[i-1] + " to " + days[i]
volatility = 0
for i in range(1, len(prices)):
  volatility += abs(prices[i] - prices[i-1])
print("Maximum price:", maximum, "on", max_day)
print("Minimum price:", minimum, "on", min_day)
print("Average price:", round(average, 2))
print("Biggest single day gain:", biggest_gain)
print("Between volatility:", volatility)
```

Q42. Given a list of 20 integers (you generate or hardcode them, some duplicates), produce: (a) the set of unique numbers, (b) the list of numbers appearing more than once, (c) the list of numbers appearing EXACTLY once. Do this WITHOUT importing Counter or any other library.

```python
numbers = [
    10, 20, 30, 20, 40,
    50, 10, 60, 70, 80,
    30, 90, 100, 40, 110,
    120, 50, 130, 140, 150
]
unique_numbers = set(numbers)
frequency = {}
for num in numbers:
  if num in frequency:
    frequency[num] += 1
  else:
    frequency[num] = 1
duplicates = []
single_occurrence = []
for num in frequency:
  if frequency[num] > 1:
    duplicates.append(num)
  elif frequency[num] == 1:
    single_occurrence.append(num)
print("Unique numbers:", unique_numbers)
print("Numbers appearing more than once:", duplicates)
print("Numbers appearing exactly once:", single_occurrence)
```

Q43. Smart Palindrome Checker. Take a sentence as input. Check if it is a palindrome IGNORING: case, spaces, and all special characters (punctuation). Examples that should return TRUE: "A man, a plan, a canal: Panama", "Race car", "No 'x' in Nixon". Hint: build a cleaned version of the string first.

```python
sentence = input("Enter a sentence:")
cleaned = ""
for char in sentence:
  if char.isalnum():
    cleaned += char.lower()
if cleaned == cleaned[::-1]:
  print("Palindrome: True")
else:
  print("Palindrome: False")
```

Q45. Caesar Cipher Encryptor. Take a message and a shift value N as input. Encrypt by shifting each letter forward by N positions in the alphabet (wrapping around, Z + 1 = A). Preserve case. Keep digits, spaces, and special characters UNCHANGED. Example: "Hello, India!" with shift 3 → "Khoor, Lqgld!"

```python
message = input("Enter a message:")
n = int(input("Enter shift value:"))
encrypted = ""
n = n % 26
for char in message:
  if 'A' <= char <= 'Z':
    new_char = chr((ord(char)-ord('A')+n) % 26 + ord('A'))
    encrypted += new_char
  elif 'a' <= char <= 'z':
    new_char = chr((ord(char)-ord('a')+n) % 26 + ord('a'))
    encrypted += new_char
  else:
    encrypted += char
print("Encrypted message:", encrypted)
```

Q46. Write a function `calculate_gst(amount, gst_rate=18, gst_inclusive=False)`: - If gst_inclusive=False: amount is base, add GST on top - If gst_inclusive=True: amount already includes GST, extract it - Return a tuple: (base_amount, gst_amount, total_amount) - All values rounded to 2 decimals Test cases: calculate_gst(1000) → (1000.00, 180.00, 1180.00) calculate_gst(1180, gst_inclusive=True) → (1000.00, 180.00, 1180.00) calculate_gst(500, gst_rate=5) → (500.00, 25.00, 525.00)

```python
def calculate_gst(amount, gst_rate = 18, gst_inclusive = False):
  if gst_inclusive == False:
    base_amount = amount
    gst_amount = base_amount * gst_rate / 100
    total_amount = base_amount + gst_amount
  else:
    base_amount = amount / (1 + gst_rate / 100)
    gst_amount = amount - base_amount
    total_amount = amount
  return(
      round(base_amount, 2),
      round(gst_amount, 2),
      round(total_amount, 2)
  )
print(calculate_gst(1000))
print(calculate_gst(1100, gst_inclusive = True))
print(calculate_gst(500, gst_rate = 5))
```

Q47. Write a RECURSIVE function `power(base, exp)` that computes base raised to exp, WITHOUT using the `**` operator and WITHOUT using `pow()`. Handle exp = 0 (return 1) and negative exp (return 1/power(base, -exp)). Test with: power(2, 10), power(5, 0), power(3, -2).

```python
def power(base, exp):
  if exp == 0:
    return 1
  if exp < 0:
    return 1 / power(base, -exp)
  return base * power(base, exp-1)
print("power(2, 10):", power(2, 10))
print("power(5, 0):", power(5, 0))
print("power(3, -2):", power(3, -2))
```

Q49. Zomato Weekly Sales Analytics. Create a NumPy array of 7-day Zomato Bengaluru sales (in ₹): sales = np.array([12000, 18500, 14200, 22100, 19800, 35200, 28400]) # Mon Tue Wed Thu Fri Sat Sun
Without using loops, compute and print: (a) Total weekly sales (b) Daily average (c) The day (Mon/Tue/...) with maximum sales (d) Number of days with sales above the weekly average (e) Percentage growth from Monday to Sunday (f) An array of day-on-day % changes (length 6)

```python

import numpy as np
sales = np.array([12000, 18500, 14200, 22100, 19800, 35200, 28400])
days = np.array(["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])
total_sales = np.sum(sales)
print("Total weekly sales: ", total_sales)
average = np.mean(sales)
print("Daily average: ", round(average, 2))
max_index = np.argmax(sales)
print("Maximum sales day:", days[max_index])
print("Maximum sales: ", sales[max_index])
above_average = np.sum(sales > average)
print("Days above average:", above_average)
growth = ((sales[-1] - sales[0]) / sales[0]) * 100
print("Percentage growth:", round(growth, 2), "%")
changes = (np.diff(sales) / sales[:-1]) * 100
print("Day-on-day percentage changes:")
print(np.round(changes, 2))
```
